# 02 Bivariate Analysis
This notebook visualizes relationships between engineered features and retention outcomes for the three project hypotheses. It is intentionally descriptive only and does not perform any statistical hypothesis testing.

## Scope
- Hypothesis 1: delivery reliability and retention
- Hypothesis 2: promotion behavior and retention
- Hypothesis 3: cross-category / cross-service usage and retention
- Retention is shown using observed `30-day`, `60-day`, and `90-day` return rates

```python
from pathlib import Path
import csv

data_path = Path('../data/processed/customer_level_analysis_dataset.csv')
with data_path.open(newline='', encoding='utf-8') as f:
    rows = list(csv.DictReader(f))
print(f'Rows: {len(rows)} | Columns: {len(rows[0])}')
```
