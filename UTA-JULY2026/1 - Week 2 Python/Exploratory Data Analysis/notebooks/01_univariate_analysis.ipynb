```python
# Distribution of customers by city
# Pre-rendered in this notebook for quick review.

from collections import Counter

city_counts = Counter(row["city"] for row in rows)
total_customers = sum(city_counts.values())

for city, count in city_counts.most_common(12):
    print(f"{city:15s} {count:5d} {count/total_customers:8.2%}")
```
