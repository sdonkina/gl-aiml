```python
# Start with an empty list and keep adding product names until the user enters "exit"
products = []

while True:
    product_name = input("Enter a product name (or type exit to stop): ")

    if product_name == "exit":
        break

    products.append(product_name)

print(products)
```

```python
# Start with an empty list and keep adding positive prices until the user enters 0
prices = []

while True:
    price = float(input("Enter a price (0 to stop): "))

    if price == 0:
        break

    if price < 0:
        print("Negative prices are not added.")
        continue

    prices.append(price)

print(prices)
```

```python
# Start with an empty list and keep adding positive prices until the user enters 0
prices = []

while True:
    user_input = input("Enter a price (0 to stop): ")

    try:
        price = float(user_input)
    except ValueError:
        print("Invalid input. Please enter a number.")
        continue

    if price == 0:
        break

    if price < 0:
        print("Negative prices are not added.")
        continue

    # Completed Logic
    prices.append(price)

print(prices)
```

```python
def calculate_final_price(price1, price2):
    total_price = price1 + price2

    if total_price > 1000:
        discount = total_price * 0.1
    else:
        discount = 0

    final_price = total_price - discount
    return final_price

# customer 1
final_1 = calculate_final_price(499.99, 899.99)
print(final_1)

# customer 2
# Completed based on the pattern above
final_2 = calculate_final_price(499.99, 199.99)
print(final_2)
```
